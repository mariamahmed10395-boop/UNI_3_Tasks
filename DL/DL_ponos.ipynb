```python
#Task__4 ponus

def downsample_temperature_data(sensor_data):

    assert sensor_data.shape == (100, 500, 500), \
        "Input data must be of shape (100, 500, 500)"

    # Divide the rows into non-overlapping blocks of 5
    downsampled_data = sensor_data.unfold(1, 5, 5)
    # Shape: (100, 100, 500, 5)

    # Divide the columns into non-overlapping blocks of 5
    downsampled_data = downsampled_data.unfold(2, 5, 5)
    # Shape: (100, 100, 100, 5, 5)

    # Calculate the average of each 5x5 block
    downsampled_data = downsampled_data.mean(dim=(-1, -2))

    return downsampled_data
```

```python
def downsample_temperature_data_view(sensor_data):

    assert sensor_data.shape == (100, 500, 500), \
        "Input data must be of shape (100, 500, 500)"

    # Divide each 500x500 image into 100x100 blocks of size 5x5
    downsampled_data = sensor_data.contiguous().view(100, 100, 5, 100, 5)
    # Shape: (100, 100, 5, 100, 5)

    # Calculate the average of each 5x5 block
    downsampled_data = downsampled_data.mean(dim=(2, 4))

    return downsampled_data


```

```python
 #Solution 1: Using unfold()

result_unfold = downsample_temperature_data(sensor_data)

#Solution 2: Using view()

result_view = downsample_temperature_data_view(sensor_data)

#Check the output shapes

print("Unfold shape:", result_unfold.shape)

print ("View shape:", result_view.shape)

#Compare the two results

print("Are the results equal?", torch.allclose(result_unfold, result_view))
```
