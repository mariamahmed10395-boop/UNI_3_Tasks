```python
import torch
from torch.utils.data import Dataset
num_samples=input("Enter num_sportamples?")
num_bits=input("Enter num_bits?")
class RandomBinaryDataSet(Dataset):
    
    def __init__(self,num_samples, num_bits ):
       super().__init__()
       self.data =torch.randint(0,2,(num_samples, num_bits ))
     

    def __len__(self):

     return len(self.data)

    def __getitem__(self,idx):    
       x=self.data[idx]
       y=torch.sum(x).item()
       return x,y
dataset=RandomBinaryDataSet(10,5)    
for i in range(5):
  x,y =dataset[i]
  print("input,",x,"label",y)       

```
