### Task 1 - Creating tensors
Create a 3x4 empty tensor, a 2x3 tensor of zeros, a 2x3 tensor of ones, and a 2x3 random tensor using seed 1729. Print the shape and dtype of each.

```python
# your code here
```

```python
# Solution
torch.manual_seed(1729)
tensors = {'empty': torch.empty(3, 4), 'zeros': torch.zeros(2, 3), 'ones': torch.ones(2, 3), 'random': torch.rand(2, 3)}
for name, tensor in tensors.items():
    print(name, tensor.shape, tensor.dtype)
```

### Task 2 - From data and data types

```python
# your code here
```

```python
# Solution
from_list = torch.tensor([[1, -1], [-1, 1]])
from_numpy = torch.from_numpy(np.array([[1, -1], [-1, 1]]))
print(torch.zeros_like(from_list))
print(from_list.float().dtype, from_list.long().dtype, from_numpy.dtype)
```

### Task 4 - Shape operations
Create the numbers 0 to 24 as a 5x5 tensor. Print the size of dim 0 and the total number of elements. Reshape it to (25,), then add a dimension at position 0 and remove it again.

```python
# your code here
```

```python
# Solution
grid = torch.arange(25).view(5, 5)
print(grid.size(0), grid.numel())
flat = grid.reshape(25)
expanded = flat.unsqueeze(0)
print(flat.shape, expanded.shape, expanded.squeeze(0).shape)
```

### Task 7 - Matrix multiplication and transpose
Multiply a 4x3 tensor of ones by its transpose using `@` and `matmul`. Then transpose a 2x3 tensor with `transpose(0, 1)`.

```python
# your code here
```

```python
# Solution
ones = torch.ones(4, 3)
print(torch.equal(ones @ ones.T, ones.matmul(ones.T)), (ones @ ones.T).shape)
print(torch.zeros(2, 3).transpose(0, 1).shape)
```

### Task 8 - Mean and standard deviation
For `m = [[1., 1.], [2., 2.], [3., 3.], [4., 4.]]`, print the overall mean, the mean along dim 0 and dim 1, and the std along dim 0. Why does it fail if `m` holds integers?

```python
# your code here
```

```python
# Solution - mean and std need a floating point dtype
m = torch.tensor([[1., 1.], [2., 2.], [3., 3.], [4., 4.]])
print(m.mean(), m.mean(0), m.mean(1), m.std(0), sep='\n')
```

### Task 9 - Concatenation
Concatenate tensors of zeros with shapes (2, 1, 3), (2, 3, 3) and (2, 2, 3) along dim 1. What is the result shape? Why does dim 0 fail?

```python
# your code here
```

```python
# Solution - all dims except the cat dim must match
parts = [torch.zeros(2, size, 3) for size in (1, 3, 2)]
print(torch.cat(parts, dim=1).shape)
```

### Task 10 - In-place operations and copying
Create `a` as a 4x3 tensor full of 3. Call `a.add(a)` and print `a`, then call `a.add_(a)` and print `a`. Finally show that `b = a` shares data while `a.clone()` does not.

```python
# your code here
```

```python
# Solution
a = torch.full((4, 3), 3.)
a.add(a)
print(a[0])
a.add_(a)
print(a[0])
label, copy = a, a.clone()
a[0, 0] = 100
print(label[0, 0], copy[0, 0])
```
