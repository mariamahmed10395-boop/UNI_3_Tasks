```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("cat.jpg", 0)

_, global_thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

# عرض النتيجة
plt.imshow(global_thresh, cmap="gray")
plt.title("Global Thresholding")
plt.axis("off")
plt.show()
```

```python

_, otsu_thresh = cv2.threshold(
    img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# عرض النتيجة
plt.imshow(otsu_thresh, cmap="gray")
plt.title("Otsu Threshold")
plt.axis("off")
plt.show()
```

```python



thresh = cv2.adaptiveThreshold(
    img, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2
)

# عرض النتيجة
plt.imshow(thresh, cmap="gray")
plt.title("Adaptive Thresholding")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("cat.jpg", 0)

_, global_thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)


_, otsu_thresh = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

adaptive_thresh = cv2.adaptiveThreshold(
    img, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2
)

# 5. عرض النتائج باستخدام Matplotlib لمقارنتها جنباً إلى جنب
plt.figure(figsize=(12, 6))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original (Grayscale)")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(global_thresh, cmap="gray")
plt.title("1. Global Thresholding (127)")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(otsu_thresh, cmap="gray")
plt.title("2. Automatic (Otsu) Thresholding")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(adaptive_thresh, cmap="gray")
plt.title("3. Adaptive Thresholding")
plt.axis("off")

plt.tight_layout()
plt.show()
```
